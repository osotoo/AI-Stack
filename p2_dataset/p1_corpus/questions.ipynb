18. 抓不到。replace 只认半角空格，全角空格不等于半角空格。判空用 strip 才安全

21. 整行判等剩 24 条，按 text 判等剩 23 条。差的就是跨来源重复 id=25
22. 返回 NaN。布尔筛选把 NaN 当 False，筛选不报错、含"发票"照常是 3 条——但这是侥幸，严谨写法是 str.contains("发票", na=False)
23. 开放题。本仓库立场：指令对构造阶段按 text 判等，id=10 留、id=25 删——同一句话教两遍没有增量价值
